# Medical Symptom Triage & Risk Predictor

### 1. Introduction

Healthcare is an important part of everyday life. However, some people have difficulty determining how serious their symptoms are and what type of medical attention they need.

The Medical Symptom Triage & Risk Predictor is an AI-based educational system designed to evaluate patient information, including symptoms and vital signs, and classify possible health risks.

The system combines Tabular Classification and a Rule-Based Expert System to generate a suggested triage level: Green (Low Risk), Yellow (Moderate Risk), or Red (High Risk).

It also provides a suggested medical specialist based on the patient's reported symptoms.

### 2. Problem Statement

Patients may have difficulty understanding the seriousness of their symptoms and identifying which type of medical specialist they should consult.

This project aims to explore how artificial intelligence can assist with preliminary health-risk classification.

The system will address the following problems:

1. Difficulty identifying the urgency of reported symptoms.
2. Lack of immediate guidance about appropriate medical specialists.
3. Possibility of incorrect health-risk classification.
4. Potential bias in medical datasets.
5. Limitations and ethical responsibilities of AI in healthcare.

### 3. Project Objectives

The main objective is to develop an AI-based prototype that classifies patient health risks using symptoms and vital signs.

Specific objectives:

1. Accept patient information, including symptoms and vital signs.
2. Apply tabular classification to evaluate patient data.
3. Use rule-based conditions to support risk classification.
4. Generate Green, Yellow, or Red triage recommendations.
5. Suggest an appropriate medical specialist.
6. Demonstrate the importance of accuracy, fairness, and patient safety in AI healthcare applications.

### 4. Proposed AI Technique

The project will use a hybrid AI approach consisting of:

**A. Tabular Classification**

This technique analyzes structured patient information, such as age, blood pressure, and other available health indicators, to classify health-risk categories.

**B. Rule-Based Expert System**

This component uses predefined medical rules to evaluate specific symptoms and vital signs.

The rules will be reviewed against appropriate medical guidance and designed conservatively to prioritize patient safety.

Combining these techniques allows the system to use both data-based classification and predefined safety rules.

### 5. Dataset Description

The project may use one of the following datasets:

- CDC Health Indicators Dataset.
- Kaggle Heart Disease Dataset.
- Kaggle Diabetes Dataset.

The selected dataset will provide structured health information for developing and testing the classification model.

The group will examine the dataset's features, quality, limitations, and suitability before implementation.

### 6. Expected System Output

The system will accept patient information and produce:

**GREEN – LOW RISK**
- No high-priority warning identified by the prototype.
- General recommendation to seek appropriate medical advice when needed.

**YELLOW – MODERATE RISK**
- Additional medical assessment is recommended.
- Suggestion to consult an appropriate healthcare professional.

**RED – HIGH RISK**
- Possible urgent health concern.
- Recommendation to seek immediate medical assessment or emergency care.

The system will also display a suggested medical specialist based on the reported symptoms.

### 7. Scope and Limitations

The project focuses on developing an educational AI prototype for preliminary health-risk classification.

Its scope includes:
- Patient information input.
- Symptom and vital-sign evaluation.
- Tabular classification.
- Rule-based evaluation.
- Triage-level recommendations.
- Suggested medical specialists.

Limitations:
- The system cannot provide a medical diagnosis.
- Predictions depend on the quality of the dataset.
- The system may produce incorrect classifications.
- The model may not perform equally well across different demographic groups.
- The prototype cannot replace doctors or emergency medical services.

### 8. Expected Outcome

At the end of the project, the group expects to develop a functional prototype that demonstrates how tabular classification and rule-based systems can work together to evaluate patient information.

The system will demonstrate basic AI applications in healthcare while highlighting the importance of accuracy, fairness, privacy, and patient safety.

# ETHICAL CONSIDERATIONS
## Medical Symptom Triage & Risk Predictor – Group 7

The Medical Symptom Triage & Risk Predictor involves health-related information, so ethical considerations must be prioritized throughout its development.

**1. Patient Safety**

Incorrect predictions may affect a person's healthcare decisions. The system must clearly communicate that its results are only preliminary recommendations and not medical diagnoses.

**2. False Negatives**

A false negative occurs when the system incorrectly classifies a patient with a serious health concern as having a low risk. This could delay necessary medical attention. Therefore, safety rules must prioritize urgent warning signs and recommend professional assessment when necessary.

**3. AI Responsibility and Liability**

Artificial intelligence should not replace medical professionals. The developers must acknowledge the limitations of the system and avoid presenting its predictions as guaranteed medical conclusions.

**4. Demographic Bias**

Medical datasets may contain unequal representation of different age groups, genders, and populations. This may affect prediction accuracy. The group must examine dataset limitations and evaluate model performance across available demographic categories.

**5. Data Privacy and Confidentiality**

Patient information must be handled responsibly. Personal identifying information should not be collected unnecessarily. Any data used for testing and development should be appropriately protected.

**Conclusion**

The project aims to demonstrate the potential benefits of AI in healthcare while recognizing that patient safety, fairness, privacy, and professional medical judgment must always remain priorities.
